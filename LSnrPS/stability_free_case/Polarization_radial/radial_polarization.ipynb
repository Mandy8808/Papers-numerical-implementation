```python
s01 = np.load("background_data_profiles_free_case/sigma_0_1_rad.npy", allow_pickle=True)
s01n1 = np.load("background_data_profiles_free_case/sigma_0_1_radn1.npy", allow_pickle=True)
s01n2 = np.load("background_data_profiles_free_case/sigma_0_1_radn2.npy", allow_pickle=True)

Ei1, rDnew1, sDnew1, dsDnew1, uDnew1, duDnew1 = s01
Ei1n1, rDnew1n1, sDnew1n1, dsDnew1n1, uDnew1n1, duDnew1n1 = s01n1
Ei1n2, rDnew1n2, sDnew1n2, dsDnew1n2, uDnew1n2, duDnew1n2 = s01n2
```

```python
fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(13, 5.5),
                       sharex=True, sharey=False,
                       gridspec_kw=dict(hspace=0.1, wspace=.17) # espacio vertical y horizontal entre los subplots
                      )

# columnas nodos 0, 1, 2 y filas: f(r), g(r)

ax[1][0].plot(rDnew1, np.abs(2*(rDnew1*sDnew1)/rDnew1 + (sDnew1 + rDnew1*dsDnew1)))
ax[1][1].plot(rDnew1n1, np.abs(2*(rDnew1n1 * sDnew1n1)/rDnew1n1 + (sDnew1n1 + rDnew1n1*dsDnew1n1)))
ax[1][2].plot(rDnew1n2, np.abs(2*(rDnew1n2*sDnew1n2)/rDnew1n2 + (sDnew1n2 + rDnew1n2*dsDnew1n2)))


ax[0][0].plot(rDnew1, np.abs(rDnew1*sDnew1))
ax[0][1].plot(rDnew1n1, np.abs(rDnew1n1*sDnew1n1))
ax[0][2].plot(rDnew1n2, np.abs(rDnew1n2*sDnew1n2))

for i in ax[1]:
   i.set_xlim(0, 20)
   i.set_xlabel(r"$r$")
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.4, 1.1)
   i.set_ylim(0.0, 3.1)


for nodo, i in enumerate(ax[0]):
   i.text(5., 0.95, r"$n=%d$"%nodo, fontsize=12)
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.6, 3.1)
   i.set_ylim(0.0, 1.1)

ax[1][0].set_ylabel(r"$|A_0|$")#(r"$-i 2^{1/2}\mu^{3/2}f(r)$")
ax[0][0].set_ylabel(r"$|A_i|$") # (r"$\sqrt{2\mu} g(r)$")

ax[0][0].set_title(r"Radial polarization", fontsize=14, loc='left')
ax[0][0].text(5., 0.85, s=r"excited state (stable)", fontsize=14)
ax[0][1].text(5., 0.85, s=r"excited state (unstable)", fontsize=14)
ax[0][2].text(5., 0.85, s=r"excited state (unstable)", fontsize=14)
```
