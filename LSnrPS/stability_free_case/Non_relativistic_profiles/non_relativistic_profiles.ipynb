## Disclaimer

In this notebook, we present the numerical profiles corresponding to stationary radially polarized and stationary linearly polarized solutions for the nodes \(n = 0, 1, 2\). In these figures, we use our numerical dimensionalization and follow the procedure described in [arXiv:2412.06901](https://arxiv.org/pdf/2412.06901) and [arXiv:2512.04376](https://arxiv.org/pdf/2512.04376), as well as in the appendix of the shared notes.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Stationary Radially Polarized Proca Star Solutions

```python
# Load data for radial polarization cases n=0,1,2
s01 = np.load("sigma_0_1_rad.npy", allow_pickle=True)
s01n1 = np.load("sigma_0_1_radn1.npy", allow_pickle=True)
s01n2 = np.load("sigma_0_1_radn2.npy", allow_pickle=True)

Ei1, rDnew1, sDnew1, dsDnew1, uDnew1, duDnew1 = s01
Ei1n1, rDnew1n1, sDnew1n1, dsDnew1n1, uDnew1n1, duDnew1n1 = s01n1
Ei1n2, rDnew1n2, sDnew1n2, dsDnew1n2, uDnew1n2, duDnew1n2 = s01n2

radius = [rDnew1, rDnew1n1, rDnew1n2]
sigma0 = [sDnew1, sDnew1n1, sDnew1n2]
uEffective = [uDnew1, uDnew1n1, uDnew1n2]
Eingevalues = [Ei1, Ei1n1, Ei1n2]
```

```python
for i in range(3):
    print(f"n={i}: E={Eingevalues[i]:.6f}, U_0={(Eingevalues[i]- uEffective[i][0]):.6f}, sigma_0={sigma0[i][0]:.6f}, ")
```

Next, we present the radial profiles of $\sigma^{(0)}$ (left panel), Newtonian potential $U=E-u_{\mathrm{eff}}$ (central panel), and mass density $\rho^{(0)}$ (right panel). Notice that the last correspond to the non-relativistic limit of $T_{00}$.

```python
col = ['#1a1919', '#2681ab', "#e30909", "#05e73e", "#e7ba05", "#a50979"]

fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 4.5),
                       sharex=False, sharey=False,
                       gridspec_kw=dict(hspace=0.2, wspace=.2)
                      )

for i in range(3):
    ax[0].plot(radius[i], radius[i]*sigma0[i], color=col[i], lw=2.5, label=r"$n=%d$"%i)
    ax[1].plot(radius[i], Eingevalues[i] - uEffective[i], color=col[i], lw=2.5, label=r"$n=%d$"%i)
    ax[2].plot(radius[i], (radius[i]*sigma0[i])**2, color=col[i], lw=2.5, label=r"$n=%d$"%i)

ax[0].legend(fontsize=12, loc='upper right', frameon=False)
ax[0].set_xlim(0, 25)
ax[2].set_xlim(0, 25)
ax[0].set_ylabel(r"$\sigma^{(0)}$")
ax[2].set_ylabel(r"$\rho^{(0)}$")
ax[1].set_ylabel(r"$U = E - u_{\mathrm{eff}}$")

ax[1].set_xlim(0, 50)

for i in ax:
   i.set_xlabel(r"$r$")
   i.hlines(0, xmin=0, xmax=125, color="black", lw=0.7, ls="--")
```

Next, we reconstructed the time and spatial components. Details are present and discusses on the shared note

```python
fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(13, 5.5),
                       sharex=True, sharey=False,
                       gridspec_kw=dict(hspace=0.1, wspace=.17) # espacio vertical y horizontal entre los subplots
                      )

# columnas nodos 0, 1, 2 y filas: f(r), g(r)
ax[0][0].plot(rDnew1, np.sqrt(2*np.pi)*(2*(rDnew1*sDnew1)/rDnew1 + (sDnew1 + rDnew1*dsDnew1)))
ax[0][1].plot(rDnew1n1, np.sqrt(2*np.pi)*(2*(rDnew1n1 * sDnew1n1)/rDnew1n1 + (sDnew1n1 + rDnew1n1*dsDnew1n1)))
ax[0][2].plot(rDnew1n2, np.sqrt(2*np.pi)*(2*(rDnew1n2*sDnew1n2)/rDnew1n2 + (sDnew1n2 + rDnew1n2*dsDnew1n2)))


ax[1][0].plot(rDnew1, np.sqrt(2*np.pi)*rDnew1*(rDnew1*sDnew1))
ax[1][1].plot(rDnew1n1, np.sqrt(2*np.pi)*rDnew1n1*(rDnew1n1*sDnew1n1))
ax[1][2].plot(rDnew1n2, np.sqrt(2*np.pi)*rDnew1n2*(rDnew1n2*sDnew1n2))

for nodo, i in enumerate(ax[0]):
   i.set_xlim(0, 20)
   i.text(6., 6.8, r"$n=%d$"%nodo, fontsize=12)
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.4, 1.1)
   i.set_ylim(-1.7, 8)

for nodo, i in enumerate(ax[1]):
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.6, 3.1)
   i.set_ylim(-5.5, 5.5)
   i.set_xlabel(r"$r$")


ax[0][0].set_ylabel(r"$h_0$")
ax[1][0].set_ylabel(r"$h_1$")

ax[0][0].set_title(r"Stationary radially polarized", fontsize=14, loc='left')
ax[0][0].text(6., 5.8, s=r"excited state (stable)", fontsize=12)
ax[0][1].text(6., 5.8, s=r"excited state (unstable)", fontsize=12)
ax[0][2].text(6., 5.8, s=r"excited state (unstable)", fontsize=12)
```

## Stationary Linearly Polarized Proca Star Solutions

Next, we present the same results as above, but in this case for the linearly polarized Proca star.

```python
s01 = np.load("sigma_0_1_LP.npy", allow_pickle=True)
s01n1 = np.load("sigma_0_1_LPn1.npy", allow_pickle=True)
s01n2 = np.load("sigma_0_1_LPn2.npy", allow_pickle=True)

Ei1, rDnew1, sDnew1, dsDnew1, uDnew1, duDnew1 = s01
Ei1n1, rDnew1n1, sDnew1n1, dsDnew1n1, uDnew1n1, duDnew1n1 = s01n1
Ei1n2, rDnew1n2, sDnew1n2, dsDnew1n2, uDnew1n2, duDnew1n2 = s01n2

radius = [rDnew1, rDnew1n1, rDnew1n2]
sigma0 = [sDnew1, sDnew1n1, sDnew1n2]
uEffective = [uDnew1, uDnew1n1, uDnew1n2]
Eingevalues = [Ei1, Ei1n1, Ei1n2]
```

```python
for i in range(3):
    print(f"n={i}: E={Eingevalues[i]:.6f}, U_0={(Eingevalues[i]- uEffective[i][0]):.6f}, sigma_0={sigma0[i][0]:.6f}, ")
```

Figure correspodent to radial profiles of $\sigma^{(0)}$ (left panel), Newtonian potential $U=E-u_{\mathrm{eff}}$ (central panel), and mass density $\rho^{(0)}$ (right panel).

```python
col = ['#1a1919', '#2681ab', "#e30909", "#05e73e", "#e7ba05", "#a50979"]

fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 4.5),
                       sharex=False, sharey=False,
                       gridspec_kw=dict(hspace=0.2, wspace=.2)
                      )

for i in range(3):
    ax[0].plot(radius[i], sigma0[i], color=col[i], lw=2.5, label=r"$n=%d$"%i)
    ax[1].plot(radius[i], Eingevalues[i] - uEffective[i], color=col[i], lw=2.5, label=r"$n=%d$"%i)
    ax[2].plot(radius[i], (sigma0[i])**2, color=col[i], lw=2.5, label=r"$n=%d$"%i)

ax[0].legend(fontsize=12, loc='upper right', frameon=False)
ax[0].set_xlim(0, 25)
ax[2].set_xlim(0, 25)
ax[0].set_ylabel(r"$\sigma^{(0)}$")
ax[1].set_ylabel(r"$U = E - u_{\mathrm{eff}}$")
ax[2].set_ylabel(r"$\rho^{(0)}$")

ax[1].set_xlim(0, 50)

for i in ax:
   i.set_xlabel(r"$r$")
   i.hlines(0, xmin=0, xmax=125, color="black", lw=0.7, ls="--")
```

Reconstructing the time and spatial components.

```python
fig, ax = plt.subplots(nrows=3, ncols=3, figsize=(13, 7.5),
                       sharex=True, sharey=False,
                       gridspec_kw=dict(hspace=0.15, wspace=.17)
                      )

ax[0][0].plot(rDnew1, np.sqrt(2*np.pi/3)*(dsDnew1))
ax[0][1].plot(rDnew1n1, np.sqrt(2*np.pi/3)*(dsDnew1n1))
ax[0][2].plot(rDnew1n2, np.sqrt(2*np.pi/3)*(dsDnew1n2))

ax[1][0].plot(rDnew1, np.sqrt(2*np.pi/3)*rDnew1*(sDnew1))
ax[1][1].plot(rDnew1n1, np.sqrt(2*np.pi/3)*rDnew1n1*(sDnew1n1))
ax[1][2].plot(rDnew1n2, np.sqrt(2*np.pi/3)*rDnew1n2*(sDnew1n2))

ax[2][0].plot(rDnew1, np.sqrt(2*np.pi/3)*rDnew1*(sDnew1)/rDnew1)
ax[2][1].plot(rDnew1n1, np.sqrt(2*np.pi/3)*rDnew1n1*(sDnew1n1)/rDnew1n1)
ax[2][2].plot(rDnew1n2, np.sqrt(2*np.pi/3)*rDnew1n2*(sDnew1n2)/rDnew1n2)

for nodo, i in enumerate(ax[0]):
   i.text(5., -0.35, r"$n=%d$"%nodo, fontsize=12)
   i.set_xlim(0, 20)
   i.set_xlabel(r"$r$")
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.25, 1.1)
   i.set_ylim(-0.7, 0.2)

for nodo, i in enumerate(ax[1]):
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.45, 0.15)
   i.set_ylim(-2, 2.3)

for i in ax[2]:
   i.set_xlim(0, 20)
   i.set_xlabel(r"$r$")
   i.hlines(0, xmin=0, xmax=20, color="black", lw=0.7, ls="--")
   #i.set_ylim(-0.25, 1.1)
   i.set_ylim(-0.45, 1.5)

ax[0][0].set_ylabel(r"$h_0$")
ax[1][0].set_ylabel(r"$h_1$")
ax[2][0].set_ylabel(r"$h_2$")

ax[0][0].set_title(r"Stationary linearly polarized", fontsize=14, loc='left')
ax[0][0].text(5., -0.45, s=r"ground state (stable)", fontsize=12)
ax[0][1].text(5., -0.45, s=r"excited state (unstable)", fontsize=12)
ax[0][2].text(5., -0.45, s=r"excited state (unstable)", fontsize=12)
```
